# The Tips Dataset (Normality Test & Log Transformation)

### Objective:
Analyse the Normality of 'tips' feature in 'tips' dataset from Seaborn. Fruther, apply logarithmic transformation to showcase how the test of normality reacts.

>Test Used: Shapiro-Wilk Test of Normality

>Performed by : Raghav Katyal on 29 September 2026

```python
#importing required libraries to perform the simulation and test
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from scipy import stats
```

```python
#Loading the dataset
df = sns.load_dataset('tips')
```

```python
#Shapiro-Wilk Test (Before Log Transformation)
w_value,p_value = stats.shapiro(df['tip'])
print(f"The W-statistic is : {w_value:.3f}")
print(f"The p-value is : {p_value:.3f}")
```

>Observation:
A W-statistic value of 0.898 and p-value of 0.000 gives aggressively strong evidence against Normality ; hence data is considered not normally distributed.

```python
#Applying Log Transformation on the 'tip' column
df['log_tip']= np.log(df['tip'])
```

```python
#Shapiro-Wilk Test (After Transformation)
w_value,p_value = stats.shapiro(df['log_tip'])
print(f"The W-statistic after log transformation is : {w_value:.3f}")
print(f"The p-value after log transformation is : {p_value:.3f}")
```

>Observation:
A W-statistic value of 0.989 and p-value of 0.056 do not show strong evidence against Normality ; hence data is considered normally distributed.

### Comparing Histograms (Before vs After Transformation)

```python
sns.set_theme(style="dark")
fig,axes = plt.subplots(nrows=1,ncols=2,figsize=(12,6))

axes[0].set_title("Distribution Before Transformation",fontweight="bold")
axes[1].set_title("Distribution After Transformation",fontweight="bold")

axes[0].set_xlabel("Tip (in $)")
axes[1].set_xlabel("Tip (in $)")

sns.histplot(data=df['tip'],ax=axes[0],kde=True)
sns.histplot(data=df['log_tip'],ax=axes[1],kde=True)
```
