# The Solar Energy Breakthrough

### The Scenario:
The Federal Network Agency states that the standard residential solar pandels produce a national historical average of 2.50 kWh of energy per day during early autumn.

A University has partnered with "The Engineering Department" to install a new,experimental light-capturing films over the solar panels on the university roof and they want to mathematically prove this new film is breakthrough.

### Objective:
Prove mathematically that The University's solar energy production average is more than the national average.

>Test Used : Z-test

>Test performed by : Raghav Katyal as on 4 October 2026

```python
import numpy as np
from scipy import stats
from statsmodels.stats.weightstats import ztest
```

```python
# University Solar Energy Data of 40 days
np.random.seed(42)
data = np.random.normal(loc=2.8,scale=0.4,size=40)
display(data)
```

```python
#Running Shapiro Test to verify the data is normal
w_value,p_value = stats.shapiro(data)
print(f"W-Statistic : {w_value:.3f}")
print(f"P-value : {p_value:.4f}")
```

>Observation : A W-value of 0.979 and a p-value of 0.6604 shows no strong evidence against Normality of data ; hence test is passed. Data is considered normally distributed.

```python
z_score,p_value = ztest(x1=data,value=2.5)
print(f"Z-score : {z_score:.3f}")
print(f"P-value : {p_value:.4f}")
```

>Observation: A Z-score of 3.527 and p-value of 0.0004 shows evidence against matching the global average ; data fails the test. Thus, it is mathematically observed that : The University's solar average are more than the national average. It is indeed breakthrough.
