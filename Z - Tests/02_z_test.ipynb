# The Laptop Battery Life Test

### Challenge:
"VoltTech" claims that their new laptop battery has an average battery life of 8 hours. Is this claim supported by the data?"

### Sample:
A quality-control team knows from previous production data that the population standard deviation is 1.2 hours. They randomly select 100 laptops from the new production batch and record their battery life.

### Objective:
Determine whether the sample provides sufficient evidence to conclude that the population mean battery life differs from 8 hours, or whether the observed difference could simply be due to random sampling variation.

### Hypotheses

**H₀ (Null Hypothesis):** The population mean battery life is 8 hours.

**H₁ (Alternative Hypothesis):** The population mean battery life is not 8 hours.

### Significance Level

α = 0.05

If p-value < 0.05 → Reject H₀  
If p-value ≥ 0.05 → Fail to reject H₀

>Test Used : Z-test

>Test performed by : Raghav Katyal as on 5 October 2026

```python
import numpy as np
from scipy import stats
```

```python
# 100 Laptop Battery Life Samples
np.random.seed(42)
data = np.random.normal(loc=8.1,scale=1.2,size=100)
display(data)
```

```python
print(f"Sample Mean Battery Life: {data.mean():.4f}")
```

```python
mu = 8
sigma = 1.2
n = len(data)

z_score = (data.mean() - mu)/(sigma/np.sqrt(n))
p_value = 2 * (1-stats.norm.cdf(abs(z_score)))

print(f"Z-Score : {z_score:.4f}")
print(f"P-value: {p_value:.4f}")
```

>Observation: The P-value is greater than 0.05, therefore we fail to reject the null hypotheses.

>Result: The test shows insufficient evidence to conclude the battery life differs from 8 hours.
