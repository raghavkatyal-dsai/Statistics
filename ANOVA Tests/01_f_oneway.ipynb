# The Employee Training Methods

### The Scenario:
A company is testing three different employee training methods.
The company wants to know whether the training method affects employees' final assessment scores.
You collect scores from 20 different employees for each training method.

### Objective:
Determine whether there is statistically significant evidence that the mean final assessment scores differ across the three training methods.

### Hypotheses

**H₀ (Null Hypothesis):** The population mean final assessment scores are equal across all three training methods.

**H₁ (Alternative Hypothesis):** At least one population mean final assessment score differs from another.

### Significance Level

α = 0.05

If p-value < 0.05 → Reject H₀  
If p-value ≥ 0.05 → Fail to reject H₀

>Test Used : One Way ANOVA
>Test Performed by Raghav Katyal as on 4 October 2026

```python
import numpy as np
from scipy import stats
```

```python
np.random.seed(42)
method1_scores = np.random.normal(loc=50,scale=10,size=20)
method2_scores = np.random.normal(loc=60,scale=10,size=20)
method3_scores = np.random.normal(loc=70,scale=10,size=20)
```

```python
#Testing Assumption of Normality using Shapiro-Wilk

#Method_1 Scores
w_stat,p_value = stats.shapiro(method1_scores)
print("The W-Statistic is: ", round(w_stat,4))
print("The P-value is: ",round(p_value,3))

#Method_2 Scores
w_stat,p_value = stats.shapiro(method2_scores)
print("\nThe W-Statistic is: ", round(w_stat,4))
print("The P-value is: ",round(p_value,3))

#Method_3 Scores
w_stat,p_value = stats.shapiro(method3_scores)
print("\nThe W-Statistic is: ", round(w_stat,4))
print("The P-value is: ",round(p_value,3))
```

>Observation: Since P-values are greater than 0.05, we fail to reject the null hypotheses of Normality.

```python
#Testing Assumption of Homogeneity of Variance using Levene's Test

stat,p_value = stats.levene(method1_scores,method2_scores,method3_scores)
print("The Levene-Statistic is: ", round(stat,4))
print("The P-value is: ",round(p_value,3))
```

>Observation: Since P-value is greater than 0.05, we fail to reject the null hypotheses of Homogeneity of Variance.

```python
print(f"Mean Method 1 : {method1_scores.mean():.3f}")
print(f"Mean Method 2 : {method2_scores.mean():.3f}")
print(f"Mean Method 3 : {method3_scores.mean():.3f}")
```

```python
f_stat,p_value = stats.f_oneway(method1_scores,method2_scores,method3_scores)
print("The F-Statistic is: ", round(f_stat,4))
print(f"The P-value is: {p_value:.2e}")
```

>Observation : Since P-value is lesser than 0.05, we reject the null hypothesis. 

>Result : There is statistically significant evidence that the mean final assessment scores are not equal across the three training methods.
