# The Teaching Methods Comparision

### Challenge:
"Does the teaching method affect students' exam scores?"

### Objective:
Determine whether there is statistically significant evidence that the mean scores differ across the three teaching methods.

### Hypotheses

**H₀ (Null Hypothesis):** The mean scores are equal across all three teaching methods.

**H₁ (Alternative Hypothesis):** At least one population mean score differs from another.

### Significance Level

α = 0.05

If p-value < 0.05 → Reject H₀  
If p-value ≥ 0.05 → Fail to reject H₀

>Test Used : One Way ANOVA
>Test Performed by Raghav Katyal as on 5 October 2026

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
method1_scores = np.random.normal(loc=55,scale=10,size=20)
method2_scores = np.random.normal(loc=55,scale=10,size=20)
method3_scores = np.random.normal(loc=56,scale=10,size=20)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(np.ones(20),y=method1_scores,label="Method 1")
plt.scatter(np.ones(20)*2,y=method2_scores,label="Method 2")
plt.scatter(np.ones(20)*3,y=method3_scores,label="Method 3")
plt.legend()
plt.title("Teaching Methods vs Student Scores")
plt.xticks([1,2,3],["Method 1","Method 2","Method 3"])
plt.ylabel("Scores")
plt.grid(alpha=0.5)
```

```python
#Testing Assumption of Normality using Shapiro-Wilk

#Method_1 Scores
w_stat,p_value = stats.shapiro(method1_scores)
print("The W-Statistic is: ", round(w_stat,4))
print("The P-value is: ",round(p_value,3))

#Method_2 Scores
w_stat,p_value = stats.shapiro(method2_scores)
print("\nThe W-Statistic is: ", round(w_stat,4))
print("The P-value is: ",round(p_value,3))

#Method_3 Scores
w_stat,p_value = stats.shapiro(method3_scores)
print("\nThe W-Statistic is: ", round(w_stat,4))
print("The P-value is: ",round(p_value,3))
```

>Observation: Since P-values are greater than 0.05, we fail to reject the null hypotheses of Normality.

```python
#Testing Assumption of Homogeneity of Variance using Levene's Test

stat,p_value = stats.levene(method1_scores,method2_scores,method3_scores)
print("The Levene-Statistic is: ", round(stat,4))
print("The P-value is: ",round(p_value,3))
```

>Observation: Since P-value is greater than 0.05, we fail to reject the null hypotheses of Homogeneity of Variance.

```python
print(f"Mean Method 1 : {method1_scores.mean():.3f}")
print(f"Mean Method 2 : {method2_scores.mean():.3f}")
print(f"Mean Method 3 : {method3_scores.mean():.3f}")
```

```python
f_stat,p_value = stats.f_oneway(method1_scores,method2_scores,method3_scores)
print("The F-Statistic is: ", round(f_stat,4))
print(f"The P-value is: {p_value:.2f}")
```

>Observation : Since P-value is greater than 0.05, we fail to reject the null hypothesis. 

>Result : There is statistically no strong evidence that proves the teaching methods affects the population mean scores.
