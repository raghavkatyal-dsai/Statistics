# The Game Completion Speed Comparision

### Challenge:
>"Does Game B take more time to complete than Game A in general?"

### The Sample:
A sample of two independent groups is taken where Group A and Group B comprising of 50 players each play Game A and Game B respectively and their game completion times are recorded.


### Objective:
Determine whether the difference between the group's game completion timings is significant than the general variation between the groups.

>Test Used : Independent T-test

>Test performed by : Raghav Katyal as on 5 October 2026

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
game_a = np.random.normal(loc=50,scale=10,size=50)
game_b= np.random.normal(loc=55,scale=8,size=50)
```

```python
plt.figure(figsize=(10,6))

plt.scatter(np.ones(50),game_a,alpha=0.7,label="Game A")
plt.scatter(np.ones(50)*2,game_b,alpha=0.7,label="Game B")
plt.xticks([1,2],["Game A","Game B"])
plt.ylabel("Completion Time (Minutes)")
plt.xlabel("Game")
plt.title("50 Players Per Game")
plt.grid(alpha=0.2)
plt.show()
```

```python
print(f"Mean of Game A Completion Timings: {game_a.mean():.3f}")
print(f"Mean of Game B Completion Timings: {game_b.mean():.3f}")
print(f"Mean Difference: {game_b.mean()-game_a.mean():.3f}")
```

>Observation : As an overview, it shows difference between the means by 7.397 minutes between Game A and Game B but Is this difference significant enough to say the population means differ or is it just random noise/variation between the groups that cause this difference?

### Hypotheses

**H₀ (Null Hypothesis):** There is no difference between the population mean completion times of Game A and Game B.

**H₁ (Alternative Hypothesis):** The population means are different. There is significant difference between the population mean completion times of Game A and Game B.

### Significance Level

α = 0.05

If p-value < 0.05 → Reject H₀  
If p-value ≥ 0.05 → Fail to reject H₀

```python
#Running Independent T-Test
t_stat,p_value = stats.ttest_ind(game_a,game_b)
print(f"T-Statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.2e}")
```

>Observation : A p-value less than 0.05 shows strong evidence against the equality of means between the groups.

>Result : There is evidence which shows the game completion times differ and on average Game B can take 7.3 minutes more completion time than Game A.
