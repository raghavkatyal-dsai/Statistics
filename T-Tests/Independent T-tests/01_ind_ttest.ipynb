# The A/B Dashboard Test

### The Scenario:
You are the Lead Data Scientist optimizing a custom data dashboard for a corporate client. The current layout is a bit cluttered. You design a new, streamlined UI layout.

You run an A/B test with 60 employees.

- Group A (Control): 30 employees use the Old Layout.

- Group B (Variant): 30 different employees use the New Layout.

You measure exactly how many seconds it takes each employee to find a specific financial metric on the screen. You want to mathematically prove to the client that your New Layout is significantly faster.

### Objective:
Determine whether the New Dashboard is significantly faster to navigate than the Old Dashboard.

>Test Used : Independent T-test

>Test performed by : Raghav Katyal as on 4 October 2026

```python
import numpy as np
from scipy import stats
```

```python
old_layout = np.random.normal(loc=45,scale=4,size=30)
new_layout = np.random.normal(loc=37,scale=4,size=30)
```

```python
#Verifying before Independent T-test : The Normality of Data
#Using Shapiro Wilk Test of Normality
print("Old Layout")
w_value,p_value = stats.shapiro(old_layout)
print(f"W-Statistic : {w_value:.3f}")
print(f"P-value : {p_value:.4f}")

print("\nNew Layout")
w_value,p_value = stats.shapiro(new_layout)
print(f"W-Statistic : {w_value:.3f}")
print(f"P-value : {p_value:.4f}")
```

>Observation : The p-values are greater than 0.05, so we fail to reject the null hypothesis of normality. There is no statistically significant evidence against normality in either group.

```python
#Running Levene's Test to Check Homogeneity of Variance
statistic,p_value = stats.levene(old_layout,new_layout)
print(f"The Levene-statistic : {statistic:.3f}")
print(f"The p-value : {p_value:.3f}")
```

>Observation : We fail to reject the null hypothesis that the group variances are equal. There is no statistically significant evidence of unequal variances.

```python
#Calculating Means of Old Layout and New Layout
print(f"The mean of navigation time in Old Layout : {old_layout.mean():.3f}")
print(f"The mean of navigation time in New Layout : {new_layout.mean():.3f}")
```

```python
#Running the Independent T-test
t_stat,p_value = stats.ttest_ind(a=old_layout,b=new_layout)
print(f"The T-statistic : {t_stat:.3f}")
print(f"The p-value : {p_value:.3f}")
```

>Observation : There is strong evidence against the null hypothesis that the two population means are equal.Therefore, we reject H₀ and conclude that the mean navigation times are statistically significantly different.

>Result : Since the mean navigation time for the New Layout is lower than that of the Old Layout, the results provide strong evidence that the New Layout is faster.
