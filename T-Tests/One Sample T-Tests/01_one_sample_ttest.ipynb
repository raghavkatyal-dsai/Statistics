# The Protein Bar Factory

### The Scenario:
A Protein Bar Factory claims every protein bar they make has average 20g of Protein. We randomly select 50 protein bars from production and measure their protein content.

### Objective:
Determine whether there is statistically significant evidence that the population mean protein content differs from the claimed 20 g.

### Hypotheses

**H₀ (Null Hypothesis):** The population mean protein content is 20 g.

**H₁ (Alternative Hypothesis):** The population mean protein content is not 20 g.

### Significance Level

α = 0.05

If p-value < 0.05 → Reject H₀  
If p-value ≥ 0.05 → Fail to reject H₀

>Test Used : One Sample T-Test
>Test Performed by Raghav Katyal as on 4 October 2026

```python
import numpy as np
from scipy import stats
```

```python
#Creating the 50 Samples Data
np.random.seed(42)
sample_data = np.random.normal(
    loc=18.5,
    scale=2,
    size=50
)
```

```python
#Checking assumptions of Normality using Shapiro Wilk Test
#Checking Sample Data
w_stat,p_value=stats.shapiro(sample_data)
print("The W-Statistic is: ", round(w_stat,4))
print("The P-value is: ",round(p_value,3))
```

>Observation: The P-value is greater than 0.05 so we fail to reject the Null Hypothesis of Normality. Hence, data can be used for One-Sample T-test.

```python
sample_mean = sample_data.mean()
claimed_mean = 20 #in grams (given as reference)

print("Sample Mean: ",sample_mean)
print("Claimed: ",claimed_mean)
print("Difference in Average : ",round(claimed_mean-sample_mean,3))
```

```python
#One-Sample T-Test (Testing the data)
t_stat,p_value = stats.ttest_1samp(sample_data,popmean=20)
print(f"The T-Statistic is: {t_stat:4f}")
print(f"The P-value is: {p_value:2e}")
```

>Observation : As the p-value is smaller than 0.05, we reject the null hypothesis. There is statistically significant evidence that the population mean protein content differs from the claimed 20 g.

>Result : Since the p-value is below 0.05, we reject H₀. There is statistically significant evidence that the population mean protein content differs from the claimed 20 g. The sample mean is approximately 1.5 g lower than the claimed value.
