# The Score Comparision

### Challenge:
>"The Sky University" claims their "Msc.Data Science" batch scored a grade of 80. Is there any evidence to this claim?"

### The Sample:
A sample of marks of 100 students from "Msc.Data Science" is taken.


### Objective:
Determine using the 100 sample student scores whether the population mean of the class is 80.

>Test Used : One Sample T-test

>Test performed by : Raghav Katyal as on 5 October 2026

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
```

```python

sample_scores = np.random.normal(loc=80,scale=5,size=100)
```

```python
plt.figure(figsize=(10,6))

plt.hist(sample_scores,color="red",edgecolor="black")
plt.title("100 Students Scores Comparision")
plt.xlabel("Scores")
plt.ylabel("Count")
plt.grid(axis="x",alpha=0.3)
```

```python
print(f"Mean of 100 Sample Scores: {sample_scores.mean():.3f}")
```

>Observation : As an overview, it shows mean sample scores as 79.481 but Is this mean score significant enough to provide evidence that the population mean is 80, or could the observed mean simply be due to random variation?

### Hypotheses

**H₀ (Null Hypothesis):** The population mean is 80.

**H₁ (Alternative Hypothesis):** The population mean is not 80.

### Significance Level

α = 0.05

If p-value < 0.05 → Reject H₀  
If p-value ≥ 0.05 → Fail to reject H₀

```python
#Running One Sample T-Test
t_stat,p_value = stats.ttest_1samp(sample_scores,popmean=80)
print(f"T-Statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.3f}")
```

>Observation : A p-value greater than 0.05 does not show strong evidence that the population mean is not 80. We fail to reject the null hypotheses.

>Result : There is insufficient evidence to prove the population mean score differs from 80. The observed sample mean of 79.481 is consistent with a population mean of 80.
