# The Teaching Methods Comparision

### Challenge:
>"Does Techning Method B really improve scores of students than when taught using Method A in general?"

### The Sample:
A sample of 100 students marks when taught using Method A and Method B respectively is taken.


### Objective:
Determine whether the observed mean paired difference is statistically significant rather than being explainable by random variation.

>Test Used : Paired T-test

>Test performed by : Raghav Katyal as on 5 October 2026

```python
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
baseline = np.random.normal(50,10,100)
scores_a = baseline + np.random.normal(2,0.5,100)
scores_b = baseline + np.random.normal(5,0.5,100)
```

```python
plt.figure(figsize=(10,6))

plt.plot(np.arange(1,101),scores_a,'o-r',label="Method A",alpha=0.4)
plt.plot(np.arange(1,101),scores_b,'o-g',label="Method B")
plt.title("Method A vs Method B (Scores Comparision)")
plt.xlabel("Student 1 to 100)")
plt.ylabel("Marks")
plt.legend()
plt.grid(axis="x",alpha=0.3)
```

```python
print(f"Mean of Method A Scores: {scores_a.mean():.3f}")
print(f"Mean of Method B Scores: {scores_b.mean():.3f}")
print(f"Mean Paired Difference: {(scores_b-scores_a).mean():.3f}")
```

>Observation : As an overview, it shows mean paired difference of scores as 3.021 between Method A and Method B but Is this mean paired difference significant enough to provide evidence that the population mean paired difference is different from zero, or could the observed difference simply be due to random variation?

### Hypotheses

**H₀ (Null Hypothesis):** The population mean paired difference of scores is zero. There is no change in scores irrespective of Methods applied.

**H₁ (Alternative Hypothesis):** The population mean paired difference is not zero. There is significant difference caused in scores because of methods applied.

### Significance Level

α = 0.05

If p-value < 0.05 → Reject H₀  
If p-value ≥ 0.05 → Fail to reject H₀

```python
#Running Paired T-Test
t_stat,p_value = stats.ttest_rel(a=scores_a,b=scores_b)
print(f"T-Statistic: {t_stat:.3f}")
print(f"P-value: {p_value:.2e}")
```

>Observation : A p-value less than 0.05 shows strong evidence that the population mean paired difference is not zero. We reject the null hypotheses.

>Result : There is evidence which shows the scores differ upon using different Methods used on same students and on average Method B improved scores by 3.021.
