# The A/B Dashboard Test

### The Scenario:
You are the Lead Data Scientist optimizing a custom data dashboard for a corporate client. The current layout is a bit cluttered. You design a new, streamlined UI layout.

You run an A/B test with 30 employees.

- Old Layout : The 30 Employees use the old layout and we record how many seconds each employee takes to find a key function.

- New Layout : The same 30 Employees now use the new layout and we record again how many seconds each employee takes to find that same key function.


### Objective:
Determine whether the New Dashboard is significantly faster to navigate than the Old Dashboard by the same employees.

>Test Used : Paired T-test

>Test performed by : Raghav Katyal as on 4 October 2026

### Hypotheses

**H₀ (Null Hypothesis):** The mean paired difference in navigation time is zero.  
There is no systematic difference between the Old and New Layout.

**H₁ (Alternative Hypothesis):** The mean paired difference in navigation time is not zero.  
There is a systematic difference between the Old and New Layout.

### Significance Level

α = 0.05

If p-value < 0.05 → Reject H₀  
If p-value ≥ 0.05 → Fail to reject H₀

```python
import numpy as np
from scipy import stats
import pandas as pd
```

```python
np.random.seed(42)
old_layout = np.random.normal(loc=45,scale=4,size=30)
improvement = np.random.normal(loc=5,scale=2,size=30)
new_layout = old_layout-improvement
df = pd.DataFrame({
    "old" : old_layout,
    "imp" : improvement,
    "new" : new_layout
})
display(df.head())

```

```python
#Verifying assumptions before Paired T-test : The Normality of Data
#Using Shapiro Wilk Test of Normality on Paired Differences (Here denoted as 'improvement')
w_value,p_value = stats.shapiro(improvement)
print(f"W-Statistic : {w_value:.3f}")
print(f"P-value : {p_value:.4f}")
```

>Observation : The p-values are greater than 0.05, so we fail to reject the null hypothesis of normality. There is no statistically significant evidence against normality in the data.

```python
#Calculating Means of Old Layout and New Layout
print(f"The mean of navigation time in Old Layout : {old_layout.mean():.3f}")
print(f"The mean of navigation time in New Layout : {new_layout.mean():.3f}")
```

```python
#Running the Paired T-test
t_stat,p_value = stats.ttest_rel(old_layout,new_layout)
print(f"The T-statistic : {t_stat:.3f}")
print(f"The p-value : {p_value:.2e}")
```

```python
mean_improvement = round(improvement.mean(),3)
mean_improvement
```

>Observation : There is strong evidence against the null hypothesis that the mean paired difference is zero. Therefore, we reject H₀ and conclude that there is a statistically significant difference in navigation time between the two layouts. The New Layout reduces navigation time by an average of 4.75 seconds.

>Result : Since the mean navigation time for the New Layout is lower than that of the Old Layout, the results provide strong evidence that the New Layout is faster.
