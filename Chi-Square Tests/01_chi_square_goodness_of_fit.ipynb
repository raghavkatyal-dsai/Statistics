# The Smartphone Factory (Goodness of Fit Test)

### Objective:

Analyze customer pre-order sales across 5 smartphone colors. Compare the physical observed sales against the factory's blueprint of equal demand (20% per color) to determine if the equal-production strategy should be scrapped.

> Test Used: Chi-Square Goodness of Fit Test

>Performed by : Raghav Katyal on 30 September 2026

```python
import numpy as np
from scipy import stats
```

```python
observed_sales = np.array([150, 130, 90, 80, 50])
expected_sales = np.array([100, 100, 100, 100, 100])
```

```python
chi_stat,p_value = stats.chisquare(f_obs=observed_sales,f_exp=expected_sales)
```

```python
print(f"Chi-Square Score : {chi_stat:.2f}")
print(f"P-value : {p_value:.5f}")
```

>Observation : A Chi-Square score of 64.00 and p-value score of 0.00 aggressively shows evidence against the goodness of fit ; huge mistmatch between the data is therefore observed. The strategy of equal production should be scraped.
