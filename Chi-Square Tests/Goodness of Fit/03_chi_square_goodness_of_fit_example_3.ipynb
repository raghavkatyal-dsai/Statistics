# The E-Commerce Fraud Detection (Chi-Square Goodness-of-Fit Test)
### Objective:

Analyze 25,000 real-world e-commerce transactions across three status categories (Legitimate, Suspicious, Confirmed Fraud) to evaluate whether actual production telemetry deviates significantly from theoretical baseline proportions using the Chi-Square Goodness-of-Fit test.

>Test Used: Chi-Square Goodness of Fit Test

>Performed by : Raghav Katyal on 01 October 2026

```python
import numpy as np
from scipy import stats
```

```python
#Baseline based on historical records
# 65% transactions fall under Legitimate
# 30% fall under Suspicious
# 5% fall under Confirmed Fraud
baseline = np.array([0.65,0.30,0.05])
```

```python
#Sample Size = 25000
size=25000
expected_transactions = baseline * size
```

```python
#Expected Frequency
print(expected_transactions)
```

```python

observed_ratio = np.array([0.63,0.34,0.03])
#Observed Frequency

observed_transactions = observed_ratio * size
print(observed_transactions)
```

```python
#Running the Chi-Square Goodness of Fit Test
chi_stat,p_value = stats.chisquare(f_obs=observed_transactions,f_exp=expected_transactions)
```

```python
print(f"The Chi-Statistic is : {chi_stat:.3f}")
print(f"The P-value is : {p_value:.4f}")
```

>Observation : A Chi-Statistic of 348.718 and p-value of 0.000 clearly shows strong evidence against Goodness of Fit ; the data fails the test. Hence, it is observed that the output did not match to the baseline.
