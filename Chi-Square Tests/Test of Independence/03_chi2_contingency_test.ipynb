# The Restaurant Gender & Smoking Independence Test (Chi-Square Test of Independence)

### Objective:
Analyze the `tips` dataset from Seaborn to investigate whether the gender (`sex`) and smoking habit (`smoker`) are statistically independent or dependent variables using the Chi-Square Test of Independence.

>Test Used: Chi-Square Test of Independence

>Performed by : Raghav Katyal on 01 October 2026

```python
import pandas as pd
import seaborn as sns
from scipy import stats
```

```python
df = sns.load_dataset('tips')
```

```python
df.head()
```

```python
#Creating the 2x2 contingency table crossing 'sex' and 'smoker' columns.
contingency_table = pd.crosstab(index=df['sex'],columns=df['smoker'])
```

```python
display(contingency_table)
```

```python
#Running the Chi-Square Test of Independence
chi_stat,p_value,_,_ = stats.chi2_contingency(contingency_table)
```

```python
print(f"Chi-Square Score : {chi_stat:.2f}")
print(f"P-value : {p_value:.5f}")
```

>Observation : A Chi-Square score of 0.00 and p-value score of 1.00 shows absolutely no strong evidence against the independence of data ; data passes the independence test. Hence it is statistically observed that : Smoking Choice is independent and not bound to Sex/Gender.
