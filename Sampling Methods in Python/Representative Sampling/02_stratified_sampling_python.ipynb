# Stratified Sampling Method in Python

> Mission : A developer wants real user reviews of his newly deployed app for enhancing it further. He needs reviews for especially how the app performs on different Operating Systems. Based on requirement he wants a random sample of 3000 reviews distributed among : 60% Windows users, 25% Linux users and 15% MacOS users from the total user reviews base.

### Strategy :
1. Using Numpy to simulate a batch of 12000 reviews combining all 3 OS platforms.
2. Using Pandas, Merging "user_id" and "user_os" in one DataFrame (df)
3. Selecting random 3000 users using stratified sampling based on the ratio 60:25:15 as required for study.

### Sampling Method:
Probability Sampling : Stratified Sampling Method

>Example created & performed in python by : Raghav Katyal on 29th September 2026

```python
#importing required libraries
import numpy as np
import pandas as pd
```

```python
#batch size (Creating the population/ universe of reviews)
batch = 12000
user_id = np.arange(1,batch+1)

user_os = ["windows","linux","mac"] * 4000
np.random.shuffle(user_os)
```

```python
#merging into one Data Frame
df = pd.DataFrame({
    "user_id" : user_id,
    "user_os" : user_os
})

#requires sample for study is 3000 among 12000
sample_size = 3000
```

```python
#selecting random reviews from each OS in the 60:25:15 ratio required for study
windows_60 = df[df['user_os']=="windows"].sample(int((60/100)*sample_size))
linux_25 = df[df['user_os']=="linux"].sample(int((25/100)*sample_size))
mac_15 = df[df['user_os']=="mac"].sample(int((15/100)*sample_size))
```

```python
#inspecting the correctness in each DataFrame
display(windows_60.head(5))
display(linux_25.head(5))
display(mac_15.head(5))
```

```python
final_stratified_sample = pd.concat([windows_60,linux_25,mac_15])
```

```python
print(f"Total Reviews fetched : {len(final_stratified_sample)}")
print(f"\nThe Ratio:")
display(final_stratified_sample['user_os'].value_counts())
```

```python
print("Stratified Sampling Successfull!")
```
