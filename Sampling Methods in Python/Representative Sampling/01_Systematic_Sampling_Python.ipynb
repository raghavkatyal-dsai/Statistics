# Systematic Sampling Method in Python

> Mission : We want to inspect every 500th item from a batch of 10,000 silicon chips in a factory.

### Strategy :
1. We use Numpy to simulate a batch of silicon chips.
2. We create 10,000 chip Id's and 10,000 temperature (in Celsius) values using np.random.normal (Gaussian/Normal Distribution).
3. Using Pandas, we merge them and then draw a systematic sample from (every 500th item in the batch).
4. Then finally using Seaborn, we plot the distribution of the sample drawn.

### Sampling Method:
Representative Sampling : Systematic Sampling Method

>Example created & performed in python by : Raghav Katyal on 28th September 2026

```python
#importing required libaries for showcasing this example
import numpy as np
import pandas as pd
import seaborn as sns
```

```python
#fixing a batch to 10,000 silicon chips out of which we shall draw samples.
batch_size = 10000

#assigning ID's to every chip produced.
chip_id = np.arange(1,batch_size+1)

#simulating Gaussian distribution to get temperatures for each chip respectively (in celsius around Average 50°C with standard deviation of 5°C).
heat_celsius = np.random.normal(loc=50,scale=5,size=batch_size)
```

```python
#creating a pandas DataFrame and adding 2 features : chip-id and its heat in celsius (which we just made)
df = pd.DataFrame({
    "chip_id" : chip_id,
    "heat_celsius" : heat_celsius
})
```

```python
#inspecting the top 5 entries
display(df.head())
```

```python
#creating the automatic process of every 500th item inspection (Systematic Sampling)
sampled_chips = df.iloc[::500]
```

```python
#Starting from 1st with a systematic gap of 500 elements : we successfully ran the Systematic Sampling.
display(sampled_chips)
```

```python
#To showcase the distribution of the sample drawn from Gaussian Distribution Population.
sns.set_theme(style="dark")
sns.histplot(data=sampled_chips,x="heat_celsius",color="teal",edgecolor="black")
```
