# Purposive + Snowball Sampling Method in Python

> Mission : We want to study the experiences of "Mountain Climbers". 

### Strategy :
1. We look among friends and find "Alex" as the only friend who we know is fond of or has been to "Mountain Climbing".
2. We interview him and ask him for references of people who he know has been to "Mountain Climbing".
3. Then we interview the references of Alex and repeat the step 2 to find more references and interview them and so on.
4. If any of the references we get are already interviewed -> we skip them.

### Sampling Method:
Hybrid Sampling : Purposive/Judgmental Sampling Method + Snowball Method

>Example created & performed in python by : Raghav Katyal on 28th September 2026

```python
#creation of the simulated friends/references of everyone mentioned in the example
contact_book = {
    "Alex" : ["Ben","Chris"],
    "Ben" : ["David"], 
    "Chris" : ["Elena"],
    "David" : [],
    "Elena" : ["Frank","Ben"],
    "Frank" : []
}
```

```python
#to maintain the records of people who we have already interviewed
interviewed_people = []

#we start from the first friend and put him on waiting list (people in queue for interviewing)
waiting_list = ["Alex"]
```

```python
# Creation of function which interviews and asks people for their references
def interviewer(current_person):
    if current_person not in interviewed_people:
            print(f"Interviewing {current_person}...")
            waiting_list.extend(contact_book.get(current_person))
            print(f"{current_person} gave references to : {contact_book.get(current_person)}")
            interviewed_people.append(current_person)
            print(f"Waiting List Updated : {waiting_list}")
            print(f"{current_person} Interviewed Successfully!...")
            print("")
```

```python
#running the loop starting from Alex -> contacting his references -> contacting his references' references.
while(len(waiting_list)!=0):
    current_person = waiting_list.pop(0)
    interviewer(current_person)

    
```

```python
#confirms that everyone in the contact list has been interviewed successfully.
print(interviewed_people)
```
